# Qwen3-0.6B × stock vLLM FlexAttention backend (reference)

이 노트북은 **vLLM 내장 `AttentionBackendEnum.FLEX_ATTENTION`** 만 사용하는 reference 다.
옆 자리의 `qwen3_flexattn_attention.ipynb` (커스텀 플러그인 = `MyFlexAttnBackend`)
가 에러를 낼 때, **stock backend 자체는 정상으로 도는가** 를 먼저 확인하기 위한 베이스라인.

차이 요약:

| | reference (이 노트북) | educational (옆 노트북) |
|---|---|---|
| backend slot | `FLEX_ATTENTION` (vLLM 빌트인) | `CUSTOM` (entry point 로 register) |
| 구현체 | `vllm.v1.attention.backends.flex_attention.FlexAttentionBackend` | `flex_attn_attention_backend:MyFlexAttnBackend` |
| plugin 의존 | 없음 (`pip install -e .` 불필요) | 있음 |
| 관찰 로그 | 없음 (stock 은 educational 로그 없음) | `MyFlexAttnImpl.forward fired ...` |
| torch.compile | stock 은 활성 (compiled flex_attention) | 비활성 (raw 호출) |

**용도**: 환경 (vllm + torch + Qwen3-0.6B) 의 동작 여부를 분리 검증.
여기서 PASS 하면 → 문제는 **커스텀 backend 코드** 안에 있다고 좁힐 수 있다.
여기서 FAIL 하면 → 환경 (torch 버전 / vLLM 버전 / GPU) 자체가 문제.

## 1. 환경 확인

vLLM 0.19.x + torch >= 2.5 + CUDA GPU.

```python
import torch, vllm

print('cuda      :', torch.cuda.is_available())
print('gpu       :', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'none')
print('vllm      :', vllm.__version__)
print('torch     :', torch.__version__)

if not torch.cuda.is_available():
    raise SystemExit('이 노트북은 CUDA GPU가 필요합니다.')

assert vllm.__version__.startswith('0.19'), (
    f'vLLM 0.19.x 권장 (현재: {vllm.__version__}). '
    '다른 버전은 stock FlexAttentionBackend 시그니처가 다를 수 있음.'
)

from torch.nn.attention.flex_attention import flex_attention, BlockMask
print('flex_attention: OK')
```

## 2. Qwen3-0.6B 구조 확인

flex_attention 은 head_dim 제한이 없다. vLLM 기본 `block_size=16` OK.

```python
from transformers import AutoConfig

cfg = AutoConfig.from_pretrained('Qwen/Qwen3-0.6B')
hd = cfg.head_dim if hasattr(cfg, 'head_dim') else cfg.hidden_size // cfg.num_attention_heads
print('hidden  :', cfg.hidden_size)
print('Q heads :', cfg.num_attention_heads, '/ KV heads:', cfg.num_key_value_heads)
print('head_dim:', hd)
print('layers  :', cfg.num_hidden_layers)
```

## 3. Backend slot 확인

`AttentionBackendEnum.FLEX_ATTENTION` 가 vLLM 빌트인 `FlexAttentionBackend` 클래스 경로로
매핑되어 있는지 확인. 이 노트북은 **CUSTOM 슬롯을 건드리지 않는다**.

```python
from vllm.v1.attention.backends.registry import AttentionBackendEnum

path = AttentionBackendEnum.FLEX_ATTENTION.get_path()
print('FLEX_ATTENTION slot ->', path)
assert 'FlexAttentionBackend' in path, f'예상 외 경로: {path}'
```

## 4. LLM 로드

`attention_backend=AttentionBackendEnum.FLEX_ATTENTION` 한 줄로 stock FlexAttention 이 붙는다.
옆 노트북과 동일한 (`max_num_batched_tokens=64`) 설정으로 chunked prefill 도 트리거.

```python
from vllm import LLM, SamplingParams
from vllm.v1.attention.backends.registry import AttentionBackendEnum

llm = LLM(
    model='Qwen/Qwen3-0.6B',
    dtype='float16',
    attention_backend=AttentionBackendEnum.FLEX_ATTENTION,
    enforce_eager=True,
    max_num_seqs=4,
    max_model_len=2048,
    max_num_batched_tokens=64,
)
```

## 5. Generate

옆 노트북과 동일한 prompt 셋. 긴 prompt 가 chunked prefill 로 쪼개진다.

```python
long_prompt = (
    'In the long history of artificial intelligence research, from the early '
    'symbolic AI of the 1950s through the neural network revival of the 1980s, '
    'the deep learning breakthroughs of the 2010s, and the transformer-based '
    'large language models of the 2020s, one theme has remained constant: '
    'the answer is'
)

prompts = [
    'The capital of France is',
    long_prompt,
    'Shakespeare wrote the play',
    'Python was created by',
]
out = llm.generate(prompts, SamplingParams(temperature=0, max_tokens=16))
for i, o in enumerate(out):
    print(f'[{i}] (prompt {len(prompts[i])} chars) {o.outputs[0].text[:80]}')
```

## 6. PASS 기준

- 위 셀이 traceback 없이 끝나고
- 4개 prompt 모두 비어 있지 않은 텍스트가 출력되면

→ stock FlexAttention backend × Qwen3-0.6B × 이 환경 조합은 **건강하다**. 옆 노트북
(`qwen3_flexattn_attention.ipynb`) 의 에러는 커스텀 `MyFlexAttnBackend` 코드
안쪽에서 발생한 것으로 좁혀진다 — `flex_attn_attention_backend.py` 의 Metadata /
Builder / Impl 셋 중 하나.

반대로 이 셀이 실패하면 — 환경 자체가 문제. 점검 순서:
1. `torch.__version__` 가 2.5 이상인가
2. `vllm.__version__` 가 0.19.x 인가
3. CUDA driver 버전이 torch CUDA 버전과 호환되는가

**stock vs educational 차이**: stock `FlexAttentionBackend` 는
`torch.compile(flex_attention, fullgraph=True)` 를 사용하고,
`direct_build` 경로 (PyTorch 2.9+) 또는 `create_block_mask_compiled` 를 사용한다.
교육용 `MyFlexAttnBackend` 는 raw `flex_attention` + `create_block_mask` (비컴파일).
두 backend 의 수치 결과는 동일해야 한다.
